```python
import pickle
import numpy as np
from GridMLM_tokenizers import CSGridMLMTokenizer
```

```python
with open('data/chord_stats.pickle', 'rb') as f:
    chord_stats = pickle.load(f)
with open('data/transition_stats.pickle', 'rb') as f:
    transition_stats = pickle.load(f)
```

```python
topk = 50

transition_freqs = np.array(list(transition_stats.values()))
transition_keys = np.array(list(transition_stats.keys()))

trans_freq_sorted_idxs = np.argsort(transition_freqs)

print(trans_freq_sorted_idxs[:topk])
print(transition_freqs[trans_freq_sorted_idxs[:topk]])
print(transition_keys[trans_freq_sorted_idxs[:topk]])
```

```python
tokenizer = CSGridMLMTokenizer(
    fixed_length=80,
    quantization='4th',
    intertwine_bar_info=True,
    trim_start=False,
    use_pc_roll=True,
    use_full_range_melody=False
)
```

```python
for i in range(topk):
    hyphen_split = transition_keys[trans_freq_sorted_idxs[i]].split('-')
    chord1 = tokenizer.ids_to_tokens[int(hyphen_split[0])]
    chord2 = tokenizer.ids_to_tokens[int(hyphen_split[1])]
    print(f'{chord1} - {chord2} : {transition_freqs[trans_freq_sorted_idxs[i]]}')
```

```python
i_offset = 6500
for i in range(topk):
    hyphen_split = transition_keys[trans_freq_sorted_idxs[i + i_offset]].split('-')
    chord1 = tokenizer.ids_to_tokens[int(hyphen_split[0])]
    chord2 = tokenizer.ids_to_tokens[int(hyphen_split[1])]
    print(f'{chord1} - {chord2} : {transition_freqs[trans_freq_sorted_idxs[i + i_offset]]}')
```

```python
# find transitions that do not exist, with chords that do exist
new_trans_existing_chords = []
for c1 in chord_stats.keys():
    for c2 in chord_stats.keys():
        tmp_trans = f'{c1}-{c2}'
        if tmp_trans not in transition_keys:
            new_trans_existing_chords.append(tmp_trans)
```

```python
print(len(new_trans_existing_chords))
for i in range(10):
    tmp_trans = new_trans_existing_chords[100*i]
    print(tmp_trans)
    hyphen_split = tmp_trans.split('-')
    chord1 = tokenizer.ids_to_tokens[int(hyphen_split[0])]
    chord2 = tokenizer.ids_to_tokens[int(hyphen_split[1])]
    print(f'{chord1} - {chord2}')
```

```python
# find transitions that do not exist, with chords that do not exist
new_trans_new_chords = []
for c1 in range(tokenizer.bar_token_id+1, len(tokenizer.vocab)):
    for c2 in range(tokenizer.bar_token_id+1, len(tokenizer.vocab)):
        if c1 not in chord_stats.keys() and c2 not in chord_stats.keys():
            tmp_trans = f'{c1}-{c2}'
            if tmp_trans not in transition_keys:
                new_trans_new_chords.append(tmp_trans)
```

```python
print(len(new_trans_new_chords))
for i in range(10):
    tmp_trans = new_trans_new_chords[100*i]
    print(tmp_trans)
    hyphen_split = tmp_trans.split('-')
    chord1 = tokenizer.ids_to_tokens[int(hyphen_split[0])]
    chord2 = tokenizer.ids_to_tokens[int(hyphen_split[1])]
    print(f'{chord1} - {chord2}')
```
