```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import re

# Set global font sizes
font_size_global = 14
plt.rcParams['font.size'] = font_size_global
plt.rcParams['axes.titlesize'] = font_size_global
plt.rcParams['axes.labelsize'] = font_size_global
plt.rcParams['xtick.labelsize'] = font_size_global
plt.rcParams['ytick.labelsize'] = font_size_global
plt.rcParams['legend.fontsize'] = font_size_global
```

```python
os.makedirs('figs', exist_ok=True)
```

```python
files_to_compare = [
    'results/LoRA/graph_contra_jnhw.csv',
    'results/LoRA/bilstm_contra_jnhw.csv',
    'results/LoRA/token_bilstm_contra_jnhw.csv',
    'results/LoRA/adapter_contra_jnhw.csv',
]
```

```python
figs_dir = 'figs/LoRA'
os.makedirs(figs_dir, exist_ok=True)

dfs = []
labels = []
for path in files_to_compare:
    df = pd.read_csv(path)
    dfs.append(df)
    labels.append(path.split('/')[2].split('_')[0])

common_columns = dfs[0].columns.tolist()[2:]
for column in common_columns:
    plt.figure(figsize=(8, 5))
    for df, label in zip(dfs, labels):
        plt.plot(df[column].values, label=label)

    plt.title(column)
    plt.xlabel('Index')
    plt.ylabel(column)
    plt.grid(True)
    plt.legend()
    plt.tight_layout()

    safe_name = re.sub(r'[^0-9A-Za-z_.-]+', '_', column)
    output_path = os.path.join(figs_dir, f'{safe_name}.png')
    plt.savefig(output_path, dpi=300)
    plt.close()
```
